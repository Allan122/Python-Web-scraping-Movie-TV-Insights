## **Combine, Clean, and Filter Data**

```python
# --- Data Preparation ---
# Add a 'Type' column to each DataFrame before we combine them.
movies_df['Type'] = 'Movie'
tv_shows_df['Type'] = 'TV Show'

# Combine the two DataFrames into a single one for easier analysis.
combined_df = pd.concat([movies_df, tv_shows_df], ignore_index=True)

# Ensure the data types are correct for filtering and calculations.
combined_df['Release Year'] = pd.to_numeric(combined_df['Release Year'])
combined_df['IMDb Rating'] = pd.to_numeric(combined_df['IMDb Rating'])

print("--- Combined and Cleaned DataFrame ---")
display(combined_df.head())
```

```python
# --- a. Filter Data based on Specific Criteria (Corrected) ---

# THE FIX: Since our data's latest year is 2023, we'll define our 'current year' as 2023
# to find items from the "last 2 years" relative to our dataset (2023 and 2022).
LATEST_YEAR_IN_DATA = 2023

# 1. Only include items released in 2023 and 2022.
# 2. Only include items with an IMDb rating of 7 or higher.
filtered_df = combined_df[
    (combined_df['Release Year'] >= (LATEST_YEAR_IN_DATA - 1)) &
    (combined_df['IMDb Rating'] >= 7.0)
].copy()

print(f"\n--- Filtered Data (Year >= {LATEST_YEAR_IN_DATA - 1} and IMDb Rating >= 7.0) ---")
display(filtered_df)
```

## **Calculate Average IMDb Rating**

```python
# --- b. Data Analysis ---
# Calculate the average IMDb rating for all scraped movies and TV shows
average_imdb_rating = combined_df['IMDb Rating'].mean()

print(f"The average IMDb rating for all 100 items is: {average_imdb_rating:.2f}")
```

## **Analyzing Top Genres**

```python
# To count genres, we split the 'Genre' string by the comma,
# then use .explode() to give each genre its own row.
# Finally, .value_counts() counts the occurrences of each genre.
genre_counts = combined_df['Genre'].str.split(', ').explode().value_counts()

top_5_genres = genre_counts.head(5)

print("--- Top 5 Genres with the Highest Number of Titles ---")
print(top_5_genres)
```

## **Finding Predominant Streaming Service**

```python
# We use the same split-explode-count method for streaming services
service_counts = combined_df['Streaming Services'].str.split(', ').explode().value_counts()

# Find the service with the most offerings
top_service_name = service_counts.index[0]
top_service_count = service_counts.iloc[0]

print(f"--- Streaming Service with the Most Offerings ---")
print(f"The predominant streaming service is '{top_service_name}' with {top_service_count} titles.")

print("\n--- Top 10 Services by Number of Titles ---")
print(service_counts.head(10))
```
